```python
# Raw order data
orders = pd.DataFrame({
    "order_id": [1, 2, 3],
    "item": ["Laptop", "Mouse", "Monitor"],
    "unit_price": [50000, 500, 22000],
    "qty": [2, 10, 3],
    "tax_pct": [18, 18, 18]
})
print(orders)
```

```python

# Add calculated columns
orders["subtotal"] = orders["unit_price"] * orders["qty"]
orders["tax_amount"] = orders["subtotal"] * orders["tax_pct"] / 100
orders["total"] = orders["subtotal"] + orders["tax_amount"]

# Drop the intermediate column
orders = orders.drop(columns=["tax_pct"])

# Rename for downstream system
orders = orders.rename(columns={"item": "product_name", "qty": "quantity"})

print(orders)
```

```python
df = pd.DataFrame({
    "emp_id": [101, 102, 103, 104],
    "name": ["Alice", "Bob", "Charlie", "Diana"],
    "base_salary": [75000, 55000, 80000, 60000],
    "bonus_pct": [10, 15, 12, 8]
})
print(df)
```

```python
# Exercise 1: Add a new column bonus_amount = base_salary * bonus_pct / 100.

df["bonus_amount"] = df["base_salary"] * df["bonus_pct"] / 100

```

```python
#Exercise 2: Add a column total_salary = base_salary + bonus_amount.

df["total_salary"] = df["base_salary"] + df["bonus_amount"]

```

```python
#Exercise 3: Add a column level with value "Senior" for salary > 70000, else "Junior". (Hint: look up np.where)

import numpy as np
df["level"] = np.where(df["base_salary"] > 70000, "Senior", "Junior")

print(df)
```

```python
#Exercise 4: Rename base_salary to base_pay and bonus_pct to bonus_percentage.

df = df.rename(columns={"base_salary": "base_pay", "bonus_pct": "bonus_percentage"})
print(df)
```

```python
#Exercise 5: Drop the bonus_pct (or bonus_percentage) column.

df = df.drop(columns=["bonus_percentage"])
print(df)
```

```python
#Exercise 6: Reorder columns as: emp_id, name, level, base_salary, bonus_amount, total_salary.

df = df[["emp_id", "name", "level", "base_pay", "bonus_amount", "total_salary"]]

print(df)
```

### Data Aggregation

Aggregation is the process of combining multiple rows into a single summary row based on one or more grouping keys. Common aggregation functions include `sum()`, `mean()`, `min()`, `max()`, and `count()`.

```python
# Group by department and calculate the average base pay
avg_salary_by_dept = df.groupby('level')['base_pay'].mean().reset_index()
print(avg_salary_by_dept)
```

```python
# Group by level and calculate multiple aggregations
salary_stats_by_level = df.groupby('level').agg(
    min_base_pay=('base_pay', 'min'),
    max_base_pay=('base_pay', 'max'),
    avg_base_pay=('base_pay', 'mean'),
    count_employees=('emp_id', 'count'),
    avg_total_salary=('total_salary', 'mean')
).reset_index()
print(salary_stats_by_level)
```
